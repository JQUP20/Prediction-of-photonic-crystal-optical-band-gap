# Photonic Crystal Generation Demo

This notebook demonstrates how to generate various types of photonic crystal structures.

```python
import sys
sys.path.append('..')

import numpy as np
import matplotlib.pyplot as plt
from src.data.photonic_crystal_generator import PhotonicCrystalGenerator
from src.utils.visualization import plot_crystal_structure
```

## Initialize Generator

```python
generator = PhotonicCrystalGenerator(random_seed=42)
```

## 1D Multilayer Structure

A simple 1D photonic crystal consisting of alternating layers of two materials.

```python
crystal_1d = generator.generate_1d_multilayer(
    n1=1.5,  # Low refractive index
    n2=3.5,  # High refractive index
    num_layers=10,
    thickness_ratio=0.5,
    lattice_constant=0.5
)

print(f"Crystal type: {crystal_1d.lattice_type}")
print(f"Lattice constant: {crystal_1d.lattice_constant} μm")
print(f"Refractive indices: {crystal_1d.refractive_indices}")
print(f"Structure shape: {crystal_1d.structure.shape}")
```

```python
# Plot 1D structure
plt.figure(figsize=(12, 4))
plt.plot(crystal_1d.structure[:, 0, 0], linewidth=2)
plt.xlabel('Position (pixels)', fontsize=12)
plt.ylabel('Refractive Index', fontsize=12)
plt.title('1D Multilayer Photonic Crystal', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.show()
```

## 2D Square Lattice

A 2D photonic crystal with circular rods arranged in a square lattice.

```python
crystal_2d_square = generator.generate_2d_square_lattice(
    n_background=1.0,  # Air
    n_rod=3.4,         # Silicon
    lattice_constant=0.5,
    rod_radius=0.3,
    grid_size=64
)

print(f"Crystal type: {crystal_2d_square.lattice_type}")
print(f"Filling fraction: {crystal_2d_square.filling_fraction:.3f}")
```

```python
# Visualize 2D square lattice
fig = plot_crystal_structure(crystal_2d_square, title="2D Square Lattice Photonic Crystal")
plt.show()
```

## 2D Triangular Lattice

A 2D photonic crystal with rods in a triangular (hexagonal) lattice arrangement.

```python
crystal_2d_tri = generator.generate_2d_triangular_lattice(
    n_background=1.0,
    n_rod=2.8,
    lattice_constant=0.6,
    rod_radius=0.25,
    grid_size=64
)

fig = plot_crystal_structure(crystal_2d_tri, title="2D Triangular Lattice Photonic Crystal")
plt.show()
```

## 3D FCC Lattice

A 3D photonic crystal with spheres in a face-centered cubic arrangement.

```python
crystal_3d = generator.generate_3d_fcc_lattice(
    n_background=1.0,
    n_sphere=2.5,
    lattice_constant=0.7,
    sphere_radius=0.2,
    grid_size=32
)

print(f"Crystal type: {crystal_3d.lattice_type}")
print(f"Structure shape: {crystal_3d.structure.shape}")
```

```python
# Visualize cross-section of 3D crystal
fig = plot_crystal_structure(crystal_3d, slice_axis=2, title="3D FCC Lattice (z-slice)")
plt.show()
```

## Generate Random Dataset

Generate a dataset of random photonic crystals for training.

```python
# Generate 10 random crystals
dataset = generator.generate_dataset(num_samples=10, grid_size=32)

print(f"Generated {len(dataset)} crystals")
print("\nDataset composition:")
lattice_types = [crystal.lattice_type for crystal in dataset]
for lt in set(lattice_types):
    count = lattice_types.count(lt)
    print(f"  {lt}: {count}")
```

```python
# Visualize first 6 samples
fig, axes = plt.subplots(2, 3, figsize=(15, 10))
axes = axes.flatten()

for i, crystal in enumerate(dataset[:6]):
    if crystal.structure.shape[2] == 1:
        slice_data = crystal.structure[:, :, 0]
    else:
        slice_data = crystal.structure[:, :, crystal.structure.shape[2]//2]
    
    axes[i].imshow(slice_data, cmap='viridis', origin='lower')
    axes[i].set_title(f"{crystal.lattice_type}\na={crystal.lattice_constant:.2f}μm", fontsize=10)
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
